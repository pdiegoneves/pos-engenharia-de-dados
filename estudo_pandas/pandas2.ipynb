```python
import pandas as pd
```

```python
vendas = pd.DataFrame({
    "produto": ["notebook", "mouse", "teclado", "notebook", "mouse"],
    "preco": [4500, 120, 350, 4800, 150],
    "quant": [1, 2, 1, 1,  3],
    "uf": ["SP", "RJ", "RJ", "MG", "SP"]
})

vendas
```

```python
vendas["total_produto_reais"] = vendas["preco"] * vendas["quant"]
resumo = vendas.groupby("produto").agg(
    preco_medio=("preco", "mean"),
    total_vendido=("quant", "sum"),     
    total_vendido_reais=("total_produto_reais", "sum"),     
)

resumo
```

```python
pvt = vendas.pivot_table(
    index="produto", columns="uf", values="quant", aggfunc="sum", fill_value=0
)
pvt
```

```python
# Merge / Concat
```

```python
clientes = pd.DataFrame({
    "id": [1, 2, 3],
    "nome": ["Ana", "João", "Maria"],
})

compras = pd.DataFrame({
    "id": [1, 1, 3, 3, 3], 
    "valor": [100, 200, 50, 75, 90]
})

display(clientes)
display(compras)
```

```python
dfm = clientes.merge(compras, on="id", how="left")
display(dfm)
dfm = clientes.merge(compras, on="id", how="right")
display(dfm)
```

```python
df1 = pd.DataFrame({
    'A': ["A0", "A1"],
    'B': ["B0", "B1"]
}, index=[0, 1])
df2= pd.DataFrame({
    'A': ["A2", "A3"],
    'B': ["B2", "B3"]
}, index=[2, 3])

display(df1)
display(df2)
```

```python
concat_vertical = pd.concat([df1, df2], axis=0)
display(concat_vertical)
```

```python
df3 = pd.DataFrame({
    'C': ["C0", "C1"],
}, index=[0, 1])
concat_horizontal = pd.concat([df1, df3], axis=1)
display(concat_horizontal)
```
