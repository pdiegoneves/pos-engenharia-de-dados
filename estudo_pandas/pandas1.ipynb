```python
df = pd.DataFrame({
    "nome": ["Ana", "João", "Maria", "Pedro", "Sofia"],
    "idade": [28, 35, 22, 41, 30],
    "cidade": ["São Paulo", "Rio de Janeiro", "São Paulo", "Belo Horizonte", "Rio de Janeiro"]
})
```
