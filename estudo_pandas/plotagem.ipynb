# Gráficos para analises

```python
dados = pd.DataFrame({
    "preco": np.random.gamma(shape=2.0, scale=80.0, size=300),
    "avaliacao": np.clip(np.random.normal(4.0, 0.5, 300), 1, 5),
    "tipo": np.random.choice(["eletronico", "roupa", "alimento"], size=300, p=[0.4, 0.3, 0.3])
})
```

```python
display(dados.head(10))
# Histograma (preço)
fig, ax = plt.subplots(figsize=(7, 4))
ax.hist(dados["preco"], bins=30)
ax.set_title("Distribuição de Preço")
ax. set_xlabel ("Preço")
ax. set_ylabel ("Frequência")
fig.tight_layout()
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
ax.scatter(dados["preco"], dados["avaliacao"], alpha=0.7)
ax.set_title("Preço vs Avaliação")
ax.set_xlabel("Preço")
ax.set_ylabel("Avaliação")
fig.tight_layout()
```

```python
grupos = [dados.loc[dados["tipo"] == t, "preco"] for t in sorted(dados["tipo"].unique())]
labels = sorted(dados["tipo"].unique())

fig, ax = plt.subplots(figsize=(7, 4))
ax.boxplot(grupos, labels=labels, showfliers=True)
ax.set_title("Preço por categoria")
ax.set_xlabel("Categoria")
ax.set_ylabel("Preço")
fig.tight_layout()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 4))

# O Seaborn entende que "tipo" é a categoria e "preco" é o valor
# O parâmetro 'order' garante que fique em ordem alfabética, igual ao seu código original
sns.boxplot(
    data=dados, 
    x="tipo", 
    y="preco", 
    order=sorted(dados["tipo"].unique()), 
    ax=ax
)

ax.set_title("Preço por categoria")
ax.set_xlabel("Categoria")
ax.set_ylabel("Preço")
fig.tight_layout()
```

```python
# Cria um DataFrame de exemplo mais complexo
np.random.seed(42)
data_vis = {
    'tipo_produto': ['eletronico', 'eletronico', 'roupa', 'eletronico', 'roupa', 'alimento' ] * 10,
    'preco': np. random. randint(50, 500, 60),
    'avaliacao': np.random.uniform(2.5, 5.0, 60).round(1)
}
df_vis = pd.DataFrame(data_vis)

```

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    x='preco',
    y='avaliacao',
    data=df_vis,
    hue='tipo_produto',
    s=100,
    palette={'eletronico': 'blue', 'roupa': 'green', 'alimento': 'red'}
)
plt.title("Relação entre Preço e Avaliação por Tipo de Produto")
plt.xlabel("Preço (R$)")
plt.ylabel("Avaliação (1-5)", fontsize=12)
plt.legend(title="Tipo de Produto")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='tipo_produto', y='preco', data=df_vis)
plt.title('Distribuição de Preços por Tipo de Produto', fontsize=16)
plt.xlabel('Tipo de Produto', fontsize=12)
plt.ylabel( 'Preço (R$)', fontsize=12)
plt.show()
```
