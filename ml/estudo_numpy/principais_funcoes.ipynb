```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
arr_1d = np.array([1, 2, 3, 4, 5])
zeros = np.zeros((2,3), dtype=np.float32)
arange = np.arange(0, 10, 2)
linespace = np.linspace(0, 1, 5)
randn = np.random.randn(3,3)
```

```python
print("arr_1d\n", arr_1d)
print("zeros\n", zeros)
print("arange\n", arange)
print("linespace\n", linespace)
print("randn\n", randn)
```

```python
M = np.array([
    [10, 20, 30],
    [10, 50, 60],
    [70, 80, 90]
])

print("Duas primeiras linhas", M[:2 :])
print("Duas primeiras colunas", M[:, :2])
print("Elemento [1, 2]: ", M[1,2])
```

```python
mask = M > 50
print("Elementos > 50 (1D): ", M[mask])
```

```python
rows = [0, 2]
cols = [1, 2]
print("Fancy indexing [0,1] e [2,2]: ", M[rows, cols])
```
